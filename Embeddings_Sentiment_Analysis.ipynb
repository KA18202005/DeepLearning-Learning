```python
docs = ['go india',
		'india india',
		'hip hip hurray',
		'jeetega bhai jeetega india jeetega',
		'bharat mata ki jai',
		'kohli kohli',
		'sachin sachin',
		'dhoni dhoni',
		'modi ji ki jai',
		'inquilab zindabad']
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts(docs)
```

```python
len(tokenizer.word_index)
```

```python
sequences = tokenizer.texts_to_sequences(docs)
sequences
```

```python
from keras.utils import pad_sequences
sequences = pad_sequences(sequences,padding='post')
sequences
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding

model = Sequential()
model.add(Embedding(len(tokenizer.word_index) + 1,output_dim=2))

model.summary()
```

```python
model.compile('adam','accuracy')
```

```python
pred = model.predict(sequences)
print(pred)
```

```python
from keras.datasets import imdb
from tensorflow.keras.preprocessing.text import Tokenizer
from keras.utils import pad_sequences
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data(num_words=10000)
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=50)
X_test = pad_sequences(X_test,padding='post',maxlen=50)
```

```python
X_train.shape
```

```python
model = Sequential()
model.add(Embedding(10000, 2, input_length=50))
model.add(SimpleRNN(32,return_sequences=False))
model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['acc'])
history = model.fit(X_train, y_train,epochs=5,validation_data=(X_test,y_test))
```

```python
word_index = imdb.get_word_index()

def preprocess_text(text):
    words = text.lower().split()
    # IMDB word index starts from 1, and 0, 1, 2 are reserved for padding, start, and unknown.
    # So, we add 3 to match the dataset's indexing.
    # We also filter words based on num_words=10000 used during imdb.load_data()
    encoded = [word_index[word] + 3 for word in words if word in word_index and word_index[word] < 10000 - 3]
    # Use 2 for unknown words (since 0, 1, 2 are reserved, 2 usually means unknown)
    encoded = [2 if w not in word_index else w for w in encoded] # This line is conceptually wrong if not careful
    # Correct way to handle unknown words based on num_words=10000
    encoded_filtered = []
    for word in words:
        if word in word_index and word_index[word] < 9997: # Adjust for the 3 special tokens
            encoded_filtered.append(word_index[word] + 3)
        else:
            encoded_filtered.append(2) # Unknown word token
    return encoded_filtered

sentence = 'i am sad'
processed_sentence = preprocess_text(sentence)
padded_sentence = pad_sequences([processed_sentence], padding='post', maxlen=50)

prediction = model.predict(padded_sentence)
print(prediction)
```
