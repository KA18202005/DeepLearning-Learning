```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('/content/shortjokes.csv')
```

```python
df
```

```python
jokes_text = df['Joke'].head(1000).tolist()
print(f"Extracted {len(jokes_text)} jokes.")
print("First 5 extracted jokes:")
for i, joke in enumerate(jokes_text[:5]):
    print(f"{i+1}. {joke}")
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts(jokes_text)
```

```python
tokenizer.word_index
```

```python
len(tokenizer.word_index)
```

```python
input_sequences = []
for sentence in jokes_text:
  tokenized_sentence = tokenizer.texts_to_sequences([sentence])[0]

  for i in range(1,len(tokenized_sentence)):
    input_sequences.append(tokenized_sentence[:i+1])
```

```python
print(f"Total input sequences generated: {len(input_sequences)}")
print(f"First 5 input sequences: {input_sequences[:5]}")
```

```python
max_sequence_len = max([len(x) for x in input_sequences])
print(f"Maximum sequence length: {max_sequence_len}")
```

```python
padded_sequences = pad_sequences(input_sequences, maxlen=max_sequence_len, padding='pre')
print(f"Shape of padded sequences: {padded_sequences.shape}")
print(f"First 5 padded sequences:\n{padded_sequences[:5]}")
```

```python
X = padded_sequences[:, :-1]
y = padded_sequences[:, -1]

print(f"Shape of X (features): {X.shape}")
print(f"Shape of y (labels): {y.shape}")
print(f"First 5 X sequences:\n{X[:5]}")
print(f"First 5 y labels: {y[:5]}")
```

```python
from tensorflow.keras.utils import to_categorical
vocabulary_size = len(tokenizer.word_index) + 1
y = to_categorical(y, num_classes=vocabulary_size)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# Redefine the model with corrections
# Use correct vocabulary_size and input_length (X.shape[1])
# Set return_sequences=True for the first LSTM
# Use correct vocabulary_size for the final Dense layer's units
model = Sequential()
model.add(Embedding(input_dim=vocabulary_size, output_dim=100, input_length=X.shape[1])) # Corrected input_dim and input_length
model.add(LSTM(150, return_sequences=True)) # Set return_sequences=True for the first LSTM
model.add(LSTM(150)) # Second LSTM layer
model.add(Dense(units=vocabulary_size, activation='softmax')) # Corrected units

# Recompile the model
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

# Print model summary to confirm changes
print(model.summary())

# Fit the model
history = model.fit(X, y, epochs=100, verbose=1)
```

```python
import time
text = "why did the chicken"

for i in range(10):
  # tokenize
  token_text = tokenizer.texts_to_sequences([text])[0]
  # padding
  padded_token_text = pad_sequences([token_text], maxlen=max_sequence_len - 1, padding='pre') # Corrected maxlen
  # predict
  pos = np.argmax(model.predict(padded_token_text, verbose=0)) # Added verbose=0 to suppress output

  for word,index in tokenizer.word_index.items():
    if index == pos:
      text = text + " " + word
      print(text)
      time.sleep(2)
```
