```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
X_train[0]
```

```python
len(X_train[2])
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=50)
X_test = pad_sequences(X_test,padding='post',maxlen=50)
```

```python
X_train[0]
```

```python
model = Sequential()

model.add(SimpleRNN(32,input_shape=(50,1),return_sequences=False))
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])

model.fit(X_train,y_train,epochs=5,validation_data=(X_test,y_test))
```

```python
test_sentence = 'I am Happy'
sequences = tokenizer.texts_to_sequences([test_sentence])
padded_sequence = pad_sequences(sequences, padding='post', maxlen=50)

# Reshape the input for the SimpleRNN layer, which expects 3D input (batch_size, timesteps, features)
# In this case, features = 1 since each token is a single integer.
reshaped_sequence = padded_sequence.reshape(1, 50, 1)

prediction = model.predict(reshaped_sequence)
print(prediction)
```
